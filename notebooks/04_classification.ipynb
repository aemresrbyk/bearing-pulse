# 04 — Random Forest, 0 HP
Split **by time inside each recording**: first 70 % of each file → train, last 30 % → test. A random split puts neighbouring (almost identical) windows on both sides and inflates the score.

```python
import sys; sys.path.insert(0, '../src')
import numpy as np, matplotlib.pyplot as plt
import bearing as b
plt.rcParams['figure.dpi'] = 110
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay
X, y, src, pos = b.dataset(loads=[0])
tr, te = pos < 0.7, pos >= 0.7
clf = RandomForestClassifier(n_estimators=200, random_state=0).fit(X[tr], y[tr])
print('time split accuracy:', accuracy_score(y[te], clf.predict(X[te])))
```

```python
ConfusionMatrixDisplay.from_estimator(clf, X[te], y[te], labels=b.CLASSES, colorbar=False)
plt.title('0 HP, time split'); plt.tight_layout(); plt.savefig('../images/01-confusion-matrix.png')
```

```python
# for comparison: a random split (what many tutorials do)
from sklearn.model_selection import train_test_split
a, c, d, e = train_test_split(X, y, test_size=0.3, random_state=0, stratify=y)
print('random split accuracy:', accuracy_score(e, RandomForestClassifier(n_estimators=200, random_state=0).fit(a, d).predict(c)))
```
