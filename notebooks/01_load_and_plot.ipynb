# 01 — Load and plot CWRU data
Drive-end accelerometer, 0 HP, one file per condition.

```python
import sys; sys.path.insert(0, '../src')
import numpy as np, matplotlib.pyplot as plt
import bearing as b
plt.rcParams['figure.dpi'] = 110
```

```python
import scipy.io
m = scipy.io.loadmat(b.RAW / '97.mat')
print({k: v.shape for k, v in m.items() if not k.startswith('__')})
```

Each file holds drive-end (`DE`), fan-end (`FE`) and sometimes base (`BA`) signals plus the measured RPM. The normal file is sampled at 48 kHz, faults at 12 kHz — `b.load` resamples normal to 12 kHz.

```python
fig, axes = plt.subplots(4, 1, figsize=(9, 7), sharex=True, sharey=True)
for ax, num in zip(axes, [97, 105, 118, 130]):
    x, rpm = b.load(num)
    t = np.arange(len(x)) / b.FS
    n = int(0.1 * b.FS)
    ax.plot(t[:n], x[:n], lw=0.6)
    ax.set_title(f'{b.FILES[num][0]}  (file {num}, {rpm:.0f} rpm)', fontsize=9, loc='left')
    ax.set_ylabel('g')
axes[-1].set_xlabel('time [s]')
fig.tight_layout(); fig.savefig('../images/01-raw-signals.png')
```

Inner and outer race faults show sharp periodic impacts; the ball fault and normal signals look much quieter.
