# 03 — Windowing and features
2048-sample windows (~0.17 s at 12 kHz, ≥ 5 shaft revolutions).

```python
import sys; sys.path.insert(0, '../src')
import numpy as np, matplotlib.pyplot as plt
import bearing as b
plt.rcParams['figure.dpi'] = 110
```

```python
X, y, src, pos = b.dataset(loads=[0])
print(X.shape)
import collections; collections.Counter(y)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(11, 3))
for ax, name in zip(axes, ['kurtosis', 'crest', 'env_BPFI']):
    i = b.FEATURE_NAMES.index(name)
    ax.boxplot([X[y == c, i] for c in b.CLASSES], tick_labels=b.CLASSES)
    ax.set_title(name, fontsize=9)
fig.tight_layout(); fig.savefig('../images/01-features.png')
```

Kurtosis and crest factor capture *impulsiveness* (impacts), envelope amplitudes at the fault frequencies capture *which* component is impacting.
